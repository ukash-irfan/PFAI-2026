```python
cleaning_log = []
students = students.drop_duplicates().reset_index(drop=True)
cleaning_log.append({"student_id": "1003", "column": "whole row",
                     "original_value": "exact duplicate", "decision": "Keep first occurrence",
                     "reason": "All column values match; one redundant row removed"})
```

```python
students["program"] = students["program"].astype("string").str.strip()
students["city"] = students["city"].astype("string").str.strip()
students[["program", "city"]]
```

```python
students["city"] = students["city"].str.title()
students["city"] = students["city"].replace({"Rwp": "Rawalpindi"})
students["city"] = students["city"].fillna("Unknown")
print("cities:", students["city"].unique())
```

```python
students["student_id"] = students["student_id"].astype("string")
conversion_columns = ["attendance", "assignment", "midterm", "final"]
missing_before_conversion = students[conversion_columns].isna().sum()
```

```python
for column in conversion_columns:
    students[column] = pd.to_numeric(students[column], errors="coerce")
new_missing = students[conversion_columns].isna().sum() - missing_before_conversion
print(students.dtypes)
print("New missing values:")
print(new_missing)
```

```python
missing_before_imputation = students[conversion_columns].isna().sum().copy()
affected_ids= {}
for column in conversion_columns:
    affected_ids[column] = students.loc[students[column].isna(), "student_id"].tolist()
students[f"{column}_imputed"] = students[column].isna()
print("Affected IDs:", affected_ids)
```

```python
for column in ["assignment", "midterm", "final"]:
    median_value = students[column].median()
    students[f"{column}_imputed"] = students[column].isna()

    for student_id in affected_ids.get(column, []):
        cleaning_log.append({
            "student_id": student_id,
            "column": column,
            "original_value": "missing",
            "decision": f"Fill with column median {median_value}",
            "reason": "Classroom simulation only; retained imputation flag"
        })

    students[column] = students[column].fillna(median_value)

students[["student_id", "assignment", "midterm", "final",
          "assignment_imputed", "midterm_imputed", "final_imputed"]]
```
